```python
#Then, we need to use some code that is separate from sciencebasepy to start the OGC cloud service
#run cell as is (will define filenames and filetypes in next cell)
def manage_cloud_map_service(sb, file_name, file_type, action, item_id):
    """
    file_type: 'raster' or 'shapefile'
    action: 'create' or 'delete'
    """
    graphql_url = sb._sbSessionEx.get_graphql_url()
    requests_session = requests.session()
    sb.refresh_token()
    gql_input = {
        "itemId": item_id,
        "fileName": file_name,
        "fileType": file_type,
        "action": action
    }

    query = """
        mutation geoserverApi($input: geoserverApiInput!) {
            geoserverApi(input: $input) {
                itemId
                fileName
                action
                statusCode
            }
        }
    """

    resp = requests_session.post(
        graphql_url,
        headers=sb._sbSessionEx.get_header(),
        json={'query': query, 'variables': {'input': gql_input}}
    )
    print ('resp: {0}\n\n'.format(resp.__dict__))
    return resp
```

```python
#Create cloud service - run as is
def create_cloud_service_By_ItemID(sb, itemID):
    
    """
    # Prototype function to auto-pull file type (raster vs shapefile) and name and call publish service.
    # Assumes just one file, useful for programmatic migrate.
    "sb": A Python ScienceBase session object, e.g., sb = pysb.SbSession()
    "itemID": the item ID with single cloud file facet to publish to cloud Geoserver. 
    """
    
    item_files = sb.get_item(itemID)['files']

    cloudShapefileCt = 0
    cloudRasterCt = 0

    startServiceFileName = ''
    startServiceFileType = ''

    for iFile in item_files:
        #print(iFile['name'])

        if iFile['name'][-4:]=='.tif' and iFile['bucket'] == 'prod-is-usgs-sb-prod-publish':
            cloudRasterCt +=1
            startServiceFileName = iFile['name']
            startServiceFileType = 'raster'
            print ("A cloud raster file ('.tif') was found in the item: " + iFile['name'])

        if iFile['name'][-4:]=='.shp' and iFile['bucket'] == 'prod-is-usgs-sb-prod-publish':
            cloudShapefileCt +=1
            startServiceFileName = iFile['name']
            startServiceFileType = 'shapefile'
            print ("A cloud shapefile ('.shp') was found in the item: " + iFile['name'])
            
    if cloudRasterCt + cloudShapefileCt != 1: 
        print ("Could not identify a single cloud shapefile or TIF to process. Investigate item.")
    else:
        print ("Publishing " + startServiceFileName + " (" + startServiceFileType + ") to Cloud Geoserver...")
        manage_cloud_map_service(sb, startServiceFileName, startServiceFileType, 'create', itemID)
            
 
```
